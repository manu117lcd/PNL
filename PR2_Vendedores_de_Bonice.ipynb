### *3.1 Análisis del corpus*

```python
import re
from collections import Counter

def find_common_contractions(texts, min_freq=5):
    # Patrón para contracciones generales (incluyendo abreviaciones como "Co.", "Inc.")
    contraction_pattern = re.compile(
        r"\b[a-zA-Z]+'[a-zA-Z]+\b|\b[A-Z][a-z]*\.\b",  # Ej: "can't", "Co."
        flags=re.IGNORECASE
    )

    all_contractions = []
    for text in texts:
        all_contractions.extend(contraction_pattern.findall(text))

    # Frecuencia de contracciones
    contraction_counts = Counter(all_contractions)
    common_contractions = {k: v for k, v in contraction_counts.items() if v >= min_freq}
    return common_contractions

# Ejecutar en tu columna 'Content'
common_contractions = find_common_contractions(df['Content'])
display("Contracciones más frecuentes:", common_contractions)
```

### *3.2 Limpieza de texto más exhaustiva.*



```python
import re
import pandas as pd

# -----------------------------------
# Función principal de limpieza
# -----------------------------------
def clean_text(text):
    if not isinstance(text, str):
        return ""

    # Eliminar JSON o listas mal formateadas como "[{...}]"
    text = re.sub(r'\[.*?\]', '', text, flags=re.DOTALL)

    # Eliminar URLs completas
    text = re.sub(r'https?://\S+|www\.\S+', '', text)

    # Eliminar emails
    text = re.sub(r'\b[\w.-]+@[\w.-]+\.\w+\b', '', text)

    # Reemplazar saltos de línea por espacio
    text = re.sub(r'\n|\r', ' ', text)

    # Eliminar múltiples espacios
    text = re.sub(r'\s+', ' ', text)

    # Limpiar caracteres sueltos (ej: " i ", " t ") que pueden ser artefactos de limpieza
    text = re.sub(r'(?<!\w)\s+[a-zA-Z]\s+(?!\w)', ' ', text)

    # Convertir a minúsculas
    text = text.strip().lower()

    return text


# -----------------------------------
# Diccionario de contracciones y abreviaciones financieras encontradas
# -----------------------------------
financial_expansions = {
    r"\bcompany's\b": "company",
    r"\bassemblies?\b": "assembly",
    r"\bagenda's\b": "agenda",
    r"\bauditor's\b": "auditor",
    r"\bcma's\b": "capital market authority",
    r"\bproject's\b": "project",
    r"\bgroup's\b": "group",
    r"\bcom\b": "commercial",
    r"\bwww\b": "",
    r"\btadawulaty\b": "",
    r"\banb\b": "national bank",
    r"\bo'clock\b": "o clock",
    r"\bqi'dah\b": "qida",
    r"\bl'azurde\b": "l azurde",
    r"\bissuer's\b": "issuer",
    r"\bdubai's\b": "dubai",
    r"\bregion's\b": "region",
    r"\bdirector's\b": "director",
    r"\bbachelor's\b": "bachelor",
    r"\bboard's\b": "board",
    r"\bexchange's\b": "exchange",
    r"\bcommittee's\b": "committee",
    r"\bauthority's\b": "authority",
    r"\bshareholder's\b": "shareholder",
    r"\byear's\b": "year",
    r"\bkingdom\b": "kingdom",
    r"\blogin\b": "",
    r"\bsmartagm\b": "",
    r"\bempower\b": "",
    r"\bedelman\b": "",
    r"\bunikai\b": "",
    r"\balinma\b": "",
    r"\bspimaco\b": "",
    r"\bsaudiexchange\b": "",
    r"\badx\b": "",
    r"\bipo\b": "",
    r"\bco\b": "company",
    r"\borg\b": "organization",
    r"\bgov\b": "government",
    r"\bprotection\b": "protection",
    r"\bdfm\b": "",
    r"\balkhaleej\b": "",
    r"\barabdrill\b": "",
    r"\bsaudicable\b": "",
    r"\bfirstmills\b": "",
    r"\bqcc\b": "",
    r"\bithmaarholding\b": "",
    r"\bdr\b": "doctor",
    r"\bmr\b": "mister",
    r"\bgeneral\b": "",
    r"\bir\b": "",
    r"\bruwais\b": "",
    r"\bprotection\b": "",
    r"\bsca\b": "",
    r"\bperiod's\b": "period",
    r"\btadawul's\b": "",
    r"\buae's\b": "united arab emirates",
    r"\bdhabi's\b": "dhabi",
    r"\binvestor\.?\b": "",
    r"\bs\.?a\.?\b": "",  # Ej: SA, S.A., etc.
    r"\bp\.?\b": "",      # Párrafo u otros usos no relevantes
    r"\bu\.?\b": "",      # Unidad u otros usos no relevantes
    r"\bj\.?\b": "",      # Iniciales
    r"\bs\.?\b": "",      # Iniciales
    r"\bc\.?\b": "",      # Iniciales
    r"\bo\.?\b": "",      # Iniciales
    r"\ba\.?\b": "",      # Iniciales
    r"\bk\.?\b": "",      # Iniciales
    r"\bn\.?\b": "",      # Iniciales
    r"\bh\.?\b": "",      # Iniciales
    r"\br\.?\b": "",      # Iniciales
    r"\bno\.?\b": "",     # Número
    r"\betc\.?\b": "",    # Etcétera
}

# -----------------------------------
# Función de expansión de contracciones financieras
# -----------------------------------
def expand_financial_contractions(text):
    for contraction, expansion in financial_expansions.items():
        try:
            text = re.sub(contraction, expansion, text, flags=re.IGNORECASE)
        except Exception as e:
            print(f"Error al procesar {contraction}: {e}")
    return text.strip()


# -----------------------------------
# Aplicar todo al DataFrame
# -----------------------------------
# Asegúrate de haber cargado df previamente

# Limpieza básica + expansión de contracciones
df['Content'] = df['Content'].apply(clean_text).apply(expand_financial_contractions)
df['CompactedSummary'] = df['CompactedSummary'].apply(clean_text).apply(expand_financial_contractions)

# Opcional: eliminar filas vacías o muy cortas tras limpieza
df = df[df['Content'].apply(lambda x: len(x.split()) > 30)].reset_index(drop=True)
```

### *3.3 Conversión del texto a minúsculas y normalización de caracteres especiales (tildes, acentos, etc.).*

```python
import unicodedata
import re

def normalize_text(text):
    if not isinstance(text, str):
        return text  # Por si hay valores no-string (ej: NaN)

    # Paso 1: Convertir a minúsculas
    text = text.lower()

    # Paso 2: Normalizar caracteres unicode (descomponer tildes y eliminar diacríticos)
    text = unicodedata.normalize('NFKD', text)\
                      .encode('ASCII', 'ignore')\
                      .decode('utf-8')

    # Paso 3: Eliminar caracteres especiales residuales (opcional)
    text = re.sub(r'[^\w\s.,;:!?¿¡$-]', '', text)  # Conserva signos de puntuación básicos

    return text

# Aplicar a la columna 'Content' (y otras si es necesario)
df['Content_normalized'] = df['Content'].apply(normalize_text)

# Mostrar ejemplo antes/después
print("=== Antes (Original) ===")
print(df['Content'][0][:100])  # Primeros 100 caracteres
print("\n=== Después (Normalizado) ===")
print(df['Content_normalized'][0][:100])
```

estandariza el formato del texto para un procesamiento NLP eficiente mediante tres pasos clave:
1. Conversión a minúsculas para uniformidad
2. Eliminación de tildes y caracteres especiales mediante normalización Unicode (evitando duplicados como "café" vs. "cafe")
3. Filtrado de caracteres residuales conservando signos de puntuación relevantes para el contexto financiero (como $ o %).

Esto garantiza que el modelo interprete consistentemente palabras y números, reduciendo ruido y mejorando la precisión en tareas como resumen automático o análisis de sentimiento.

### *3.4 Tokenización*

```python
# Función para dividir en oraciones (conservando términos financieros)
def split_sentences(text):
    return re.split(r'(?<!\w\.\w.)(?<![A-Z][a-z]\.)(?<=\.|\?|\!)\s+', text)

# Función para tokenizar palabras/símbolos financieros
def tokenize_financial_text(text):
    return re.findall(
        r'\b\w+(?:\.\w+)*\b|[\$\€\£]\d+\.?\d*|\d+\.?\d*%|[.,;:!?¿¡]',
        text
    )

# Procesamiento completo
def process_text(df):
    # 1. Tokenizar por oraciones y palabras
    df['Sentences'] = df['Content'].apply(split_sentences)
    df['Tokens'] = df['Sentences'].apply(
        lambda sentences: [tokenize_financial_text(sent) for sent in sentences]
    )

    return df
```

```python
# Aplicar al DataFrame (asumiendo df['Content'] está limpio/normalizado)
df = process_text(df)

# Mostrar ejemplo tokenizado
print("=== Ejemplo de Tokens por Oración ===")
print(df['Tokens'][0][:2])  # Primeras 2 oraciones tokenizadas
```

Se optó por este método para garantizar que términos financieros clave (como símbolos monetarios $, porcentajes %, abreviaciones Q2 y nombres de empresas Inc.) se conserven íntegros durante la tokenización, evitando divisiones incorrectas que afecten el significado. El uso de expresiones regulares personalizadas permite:

Precisión en datos financieros: capturar montos ($1.5B), fechas y entidades (Apple Inc.) como unidades completas.

Eliminación selectiva de ruido: filtrar stopwords genéricas pero preservar negaciones (not) y condicionales (under), críticas en análisis financiero.

Consistencia con el preprocesamiento previo, ya que opera sobre texto ya normalizado (minúsculas, sin acentos).

Los tokens por oración mantiene la relación contextual entre palabras (útil para resúmenes).

## **5. Extracción de información e Identificación de palabras clave**

### *5.1 Identificar automáticamente elementos clave en el texto:*


 Entidades Nombradas (NER) para detectar “organizaciones”, “personas”, “lugares”, “fechas”, “cantidades monetarias”, “acciones o eventos”.
Expresiones regulares para casos concretos como montos de dinero, porcentajes, fechas en diferentes formatos.  

** Mejoras aplicadas para optimizar la extracción de entidades**

Se aplicaron dos optimizaciones principales para mejorar la precisión de la extracción:
- **Limpieza de fechas incompletas**: se eliminaron registros de fechas que no contenían información numérica suficiente, como "2024-03-", para asegurar su correcto análisis.
- **Filtrado de organizaciones irrelevantes**: se descartaron entradas genéricas o materiales como "Polypropylene" o "Company" que no representan entidades reales de interés.

Estas mejoras permiten obtener datos estructurados de manera más confiable, alineados con los objetivos del proyecto: identificar organizaciones, personas, lugares, fechas, cantidades monetarias y eventos relevantes.

**Nota**: No se detectaron personas en los textos analizados debido a que los documentos corresponden a anuncios empresariales donde típicamente se mencionan compañías, contratos y actividades, pero no individuos concretos.


```python
# ----------------------------------------------
# 1. Cargar modelo de spaCy en inglés
# ----------------------------------------------
!python -m spacy download en_core_web_lg

import spacy
import pandas as pd
import re
from collections import Counter
import matplotlib.pyplot as plt

# Cargar modelo grande de spaCy en inglés
nlp = spacy.load("en_core_web_lg")

# ----------------------------------------------
# 2. Función para extraer entidades
# ----------------------------------------------
def extraer_elementos_clave_final(texto):
    doc = nlp(texto)

    entidades = {
        "PERSONAS": [],
        "ORGANIZACIONES": [],
        "LUGARES": [],
        "FECHAS": [],
        "DINERO": [],
        "EVENTOS": [],
        "PORCENTAJES": [],
        "FECHAS_REGEX": [],
        "MONTOS_DINERO_REGEX": []
    }

    palabras_invalidas_org = ["polypropylene", "company", "previous announcement", "click here", "website", "material", "introduction"]

    for ent in doc.ents:
        if ent.label_ == "PERSON":
            if len(ent.text.split()) >= 2 and ent.text.isalpha():
                entidades["PERSONAS"].append(ent.text.strip())
        elif ent.label_ == "ORG":
            if (1 <= len(ent.text.split()) <= 6) and not any(bad_word in ent.text.lower() for bad_word in palabras_invalidas_org):
                entidades["ORGANIZACIONES"].append(ent.text.strip())
        elif ent.label_ == "GPE":
            entidades["LUGARES"].append(ent.text.strip())
        elif ent.label_ == "DATE":
            if any(char.isdigit() for char in ent.text) and len(ent.text.strip()) <= 20:
                entidades["FECHAS"].append(ent.text.strip())
        elif ent.label_ == "MONEY":
            entidades["DINERO"].append(ent.text.strip())

    # Detección manual de eventos relevantes
    eventos_detectados = re.findall(r'\b(merger|acquisition|dividend|ipo|split|shutdown|launch|appointment|maintenance)\b', texto, flags=re.IGNORECASE)
    entidades["EVENTOS"].extend([evento.lower() for evento in eventos_detectados])

    # Expresiones regulares complementarias
    regex_dinero = r'\$\s?\d{1,3}(?:,\d{3})*(?:\.\d{2})?|\d+\s?(million|billion|trillion)'
    regex_porcentajes = r'\b\d{1,3}(?:\.\d+)?%'
    regex_fechas = r'\b(?:\d{1,2}[/-]){2}\d{2,4}\b'

    entidades["MONTOS_DINERO_REGEX"].extend(re.findall(regex_dinero, texto, flags=re.IGNORECASE))
    entidades["PORCENTAJES"].extend(re.findall(regex_porcentajes, texto))
    entidades["FECHAS_REGEX"].extend(re.findall(regex_fechas, texto))

    # Eliminar duplicados
    entidades = {k: list(set(v)) for k, v in entidades.items() if v}

    return entidades

# ----------------------------------------------
# 3. Aplicar extracción de entidades
# ----------------------------------------------
columna_texto = 'Content'

df['Entities'] = df[columna_texto].dropna().apply(extraer_elementos_clave_final)

# ----------------------------------------------
# 4. Función para limpiar fechas incompletas y organizaciones irrelevantes
# ----------------------------------------------
def limpiar_entidades(entidades):
    palabras_prohibidas = ["polypropylene", "company", "previous announcement", "click here", "website", "material", "introduction"]

    entidades_limpias = {}

    for categoria, lista in entidades.items():
        if categoria == "FECHAS":
            fechas_validas = [fecha for fecha in lista if len(re.findall(r'\d', fecha)) >= 6]
            if fechas_validas:
                entidades_limpias[categoria] = fechas_validas
        elif categoria == "ORGANIZACIONES":
            org_validas = []
            for org in lista:
                texto = org.lower()
                if not any(bad_word in texto for bad_word in palabras_prohibidas) and len(org.split()) >= 2:
                    org_validas.append(org.strip())
            if org_validas:
                entidades_limpias[categoria] = org_validas
        else:
            entidades_limpias[categoria] = lista

    return entidades_limpias

# Aplicar limpieza
df['Entities_Limpias'] = df['Entities'].dropna().apply(limpiar_entidades)

# ----------------------------------------------
# 5. Guardar resultados limpios
# ----------------------------------------------
df[['Content', 'Entities_Limpias']].to_csv('noticias_procesadas_limpias.csv', index=False)
print("\nArchivo 'noticias_procesadas_limpias.csv' guardado correctamente.")

# ----------------------------------------------
# 6. Resumen de entidades detectadas
# ----------------------------------------------
conteo_entidades_limpias = Counter()

for entidades in df['Entities_Limpias'].dropna():
    for tipo, valores in entidades.items():
        conteo_entidades_limpias[tipo] += len(valores)

resumen_limpio = pd.DataFrame.from_dict(conteo_entidades_limpias, orient='index', columns=['Cantidad'])
resumen_limpio = resumen_limpio.sort_values(by='Cantidad', ascending=False)
print("\nResumen de entidades detectadas (limpio):")
display(resumen_limpio)

# Graficar
resumen_limpio.plot(kind='bar', legend=False)
plt.title('Conteo de Entidades Detectadas (limpio)')
plt.xlabel('Tipo de Entidad')
plt.ylabel('Cantidad')
plt.xticks(rotation=45)
plt.show()
```

```python
import math
from collections import defaultdict
```

```python
#Función para calcular TF (Term Frequency)
def compute_tf(document):
    """Calcula la frecuencia de términos en un solo documento."""
    tf = defaultdict(int)
    total_terms = len(document)
    for term in document:
        tf[term] += 1
    # Normalizar por la longitud del documento
    tf = {term: count / total_terms for term, count in tf.items()}
    return tf
```

```python
#Función para calcular IDF (Inverse Document Frequency)
def compute_idf(documents):
    """Calcula la frecuencia inversa de documentos para todos los términos."""
    idf = defaultdict(float)
    total_documents = len(documents)

    # Contar en cuántos documentos aparece cada término
    term_doc_count = defaultdict(int)
    for doc in documents:
        unique_terms = set(doc)
        for term in unique_terms:
            term_doc_count[term] += 1

    # Calcular IDF para cada término
    idf = {term: math.log(total_documents / (count + 1))  # +1 para evitar división por cero
           for term, count in term_doc_count.items()}
    return idf
```

```python
#Función para calcular TF-IDF
def compute_tfidf(documents):
    """Calcula TF-IDF para todos los documentos."""
    # Preprocesamiento: tokenizar documentos (asumiendo que ya están tokenizados)
    tokenized_docs = [doc.split() if isinstance(doc, str) else doc for doc in documents]

    # Calcular IDF para el corpus
    idf = compute_idf(tokenized_docs)

    # Calcular TF-IDF para cada documento
    tfidf_results = []
    for doc in tokenized_docs:
        tf = compute_tf(doc)
        tfidf = {term: tf_val * idf.get(term, 0) for term, tf_val in tf.items()}
        tfidf_results.append(tfidf)

    return tfidf_results
```

```python
#Función para obtener las top N palabras clave por documento
def get_top_keywords(tfidf_results, top_n=5):
    """Extrae las N palabras con mayor puntuación TF-IDF por documento."""
    top_keywords = []
    for doc_tfidf in tfidf_results:
        # Ordenar términos por score TF-IDF (de mayor a menor)
        sorted_terms = sorted(doc_tfidf.items(), key=lambda x: x[1], reverse=True)
        top_keywords.append([term for term, score in sorted_terms[:top_n]])
    return top_keywords
```

#### ***5.2.2RAKE***

```python
import re
from collections import defaultdict, Counter
import nltk
from nltk.corpus import stopwords

# Descargar recursos necesarios (una sola vez)
nltk.download('stopwords')

# Definir stopwords y puntuación
stop_words = set(stopwords.words('english'))
punctuation = r'[^\w\s]'

# 1. Función para generar candidatos RAKE
def generate_candidate_keywords(text):
    # Dividir en oraciones o fragmentos
    sentence_delimiters = re.compile(r'[.!?;]\s*|\n')
    phrases = sentence_delimiters.split(text)

    candidates = []
    for phrase in phrases:
        # Extraer palabras y filtrar por stopwords y puntuación
        words = re.findall(r'\b\w+(?:\.\w+)*\b', phrase.lower())
        filtered_words = [
            word for word in words
            if word not in stop_words and not re.search(punctuation, word)
        ]
        if filtered_words:  # Solo agregar si hay palabras válidas
            candidates.append(' '.join(filtered_words))
    return candidates

# 2. Función para calcular puntuaciones RAKE
def calculate_rake_scores(candidates):
    word_freq = Counter()
    word_degree = Counter()

    for phrase in candidates:
        words = phrase.split()
        phrase_length = len(words)

        for word in words:
            word_freq[word] += 1
            word_degree[word] += phrase_length - 1  # Grado = longitud frase -1

    # Calcular score individual por palabra
    word_scores = {}
    for word in word_freq:
        word_scores[word] = word_degree[word] / word_freq[word]

    # Calcular score por frase
    phrase_scores = {}
    for phrase in candidates:
        words = phrase.split()
        phrase_score = sum(word_scores.get(word, 0) for word in words)
        phrase_scores[phrase] = phrase_score

    return phrase_scores

# 3. Función principal: extraer top-N frases clave
def extract_rake_keywords(text, top_n=5):
    candidates = generate_candidate_keywords(text)
    scores = calculate_rake_scores(candidates)

    # Ordenar y seleccionar top N
    sorted_phrases = sorted(scores.items(), key=lambda x: x[1], reverse=True)
    return [phrase for phrase, score in sorted_phrases[:top_n]]

# Aplicar al DataFrame usando la columna original 'Content'
df['RAKE_Keywords'] = df['Content'].apply(lambda x: extract_rake_keywords(x, top_n=5))

# Mostrar resultados
display(df[['Subject', 'RAKE_Keywords']])
```

#### ***5.2.3 Embeddings***

**Extracción de Palabras Clave utilizando Embeddings de spaCy.**

* Se implementó un método basado en embeddings de palabras utilizando el modelo preentrenado `en_core_web_lg` de spaCy.

* Cada documento fue tokenizado eliminando signos de puntuación y palabras vacías (stopwords).

* Posteriormente, se calcularon los vectores de cada palabra y el vector promedio del documento.

* Las palabras clave fueron seleccionadas como aquellas cuyo embedding resultó más cercano al centro semántico del texto, basado en la similitud de coseno.

* Esta estrategia permitió identificar términos representativos de la temática principal de cada noticia de forma eficiente y robusta, sin depender de bibliotecas adicionales.

Para cada documento;

* Obtener las palabras que aparecen en él.

* Calcular el vector promedio del documento.

* Buscar las palabras más cercanas a ese vector promedio.

* Seleccionar las n palabras clave que representen cada noticia.



```python
# ----------------------------------------------
# 1. Cargar spaCy (ya instalado)
# ----------------------------------------------
import spacy
import pandas as pd
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

# Cargar modelo grande de inglés
nlp = spacy.load("en_core_web_lg")

# ----------------------------------------------
# 2. Preprocesar textos: tokenizar usando spaCy
# ----------------------------------------------
def preprocesar_texto_spacy(texto):
    doc = nlp(texto.lower())
    palabras = [token.text for token in doc if token.is_alpha and not token.is_stop]
    return palabras

# Aplicar preprocesamiento
df['Tokens'] = df['Content'].dropna().apply(preprocesar_texto_spacy)

# ----------------------------------------------
# 3. Función para extraer palabras clave usando spaCy
# ----------------------------------------------
def extraer_keywords_spacy(palabras, top_n=5):
    palabras_validas = [p for p in palabras if p in nlp.vocab and nlp.vocab[p].has_vector]

    if not palabras_validas:
        return []

    # Obtener vectores
    vectores = np.array([nlp.vocab[p].vector for p in palabras_validas])

    # Vector promedio del documento
    vector_doc = np.mean(vectores, axis=0).reshape(1, -1)

    # Calcular similitud coseno
    similitudes = cosine_similarity(vectores, vector_doc).flatten()

    # Seleccionar top_n palabras
    indices_top = np.argsort(similitudes)[::-1][:top_n]
    palabras_clave = [palabras_validas[i] for i in indices_top]

    return palabras_clave

# ----------------------------------------------
# 4. Aplicarlo al DataFrame
# ----------------------------------------------
df['Keywords_SpaCy_Embeddings'] = df['Tokens'].apply(lambda tokens: extraer_keywords_spacy(tokens, top_n=5))

# ----------------------------------------------
# 5. Mostrar ejemplos
# ----------------------------------------------
print("\nPalabras clave por spaCy (noticia 2):")
display(df[['Content', 'Keywords_SpaCy_Embeddings']].iloc[2])

print("\nPalabras clave por spaCy (noticia 4):")
display(df[['Content', 'Keywords_SpaCy_Embeddings']].iloc[4])

print("\nPalabras clave por spaCy (noticia 6):")
display(df[['Content', 'Keywords_SpaCy_Embeddings']].iloc[6])

# ----------------------------------------------
# 6. Guardar resultados
# ----------------------------------------------
df[['Content', 'Keywords_SpaCy_Embeddings']].to_csv('noticias_palabras_clave_spacy.csv', index=False)
print("\nArchivo 'noticias_palabras_clave_spacy.csv' guardado correctamente.")
```

### *6.1 Resumen extractivo: TextRank, Sumy, Gensim.*

#### ***6.1.1 TextRank***

```python
import re
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer, CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import networkx as nx

# --------------------------------------------
# Función auxiliar: dividir texto en bloques/oraciones
# --------------------------------------------
def split_sentences_flexible(text, chunk_size=30):
    words = text.split()
    return [' '.join(words[i:i + chunk_size]) for i in range(0, len(words), chunk_size)]

# --------------------------------------------
#  TextRank (grafo de similitud entre oraciones)
# --------------------------------------------
def summarize_by_textrank(text, top_n=3):
    if not isinstance(text, str) or not text.strip():
        return ["Texto vacío o inválido."]

    sentences = split_sentences_flexible(text.strip(), chunk_size=30)
    sentences = list(set(sentences))
    sentences = [s for s in sentences if len(s.split()) >= 3]

    if not sentences:
        return ["Texto demasiado corto o sin oraciones válidas para resumir."]
    try:
        vectorizer = TfidfVectorizer(stop_words='english')
        tfidf_matrix = vectorizer.fit_transform(sentences)
    except ValueError:
        return ["No se pudieron generar features TF-IDF (vocabulario vacío)."]

    if tfidf_matrix.shape[0] == 0:
         return ["No se pudieron generar features TF-IDF (matriz vacía)."]

    similarity_matrix = cosine_similarity(tfidf_matrix)

    if similarity_matrix.shape[0] <= 1:
         return sentences

    graph = nx.from_numpy_array(similarity_matrix)

    try:
        scores = nx.pagerank(graph)
    except nx.PowerIterationFailedConvergence:
        return sentences[:top_n] if len(sentences) >= top_n else sentences

    ranked_sentences = sorted([(scores[i], s) for i, s in enumerate(sentences)], key=lambda x: x[0], reverse=True)
    return [s for _, s in ranked_sentences[:top_n]]


# --------------------------------------------
# Función de comparación visual de resúmenes
# --------------------------------------------
def compare_summaries(title, summary1):
    print(f"\n--- COMPARACIÓN DE RESÚMENES ---\nTítulo: {title}\n")
    print("\n TextRank:")
    if isinstance(summary1, list):
        for i, sent in enumerate(summary1, 1):
            display(f"{i}. {sent}")
    else:
        print(summary1)
    print("-" * 80)

# --------------------------------------------
# Aplicar a una muestra del dataset
# --------------------------------------------
df['resumen_textrank'] = df['Content'].apply(lambda x: summarize_by_textrank(x, top_n=3))
```

```python
#Una vez instaladas las dependencias, procedemos a cargar el modelo T5 (Transformer)
from transformers import T5Tokenizer, T5ForConditionalGeneration
from tqdm import tqdm

#Si quisieramos obtener resumenes de mejor "calidad" podríamos usar el modelo "t5-large"
model_name = 't5-base'
#Ahora le pasamos a la librería que modelo queremos usar
tokenizer = T5Tokenizer.from_pretrained(model_name)
model = T5ForConditionalGeneration.from_pretrained(model_name)

#Ahora creamos una función para que poder hacer los resumenes de los artículos de nuesto dataset
#La función recibe como parámetros los textos a los que se les quiere aplicar el resumen con el modelo, también se tiene
#algunos parametros con un valor por defecto, max_long es el máximo que puede tener la entrda de texto, ya que el modelo
#T5 tiende a fallar con textos muy grandes, los otros dos parámetros tienen que ver con la longitud que queremos que
#tenga el resumen generado.
def resumir_textos_t5(textos, max_long=512, resumen_max=100, resumen_min=30, penalizacion=0.0):

  resúmenes = []
  for texto in tqdm(textos, desc="Generando resúmenes"):

      #Es necesario decirle al modelo que tipo de tarea queremos realizar ya que el modelo tiene un efoque "multitarea"
      #para traducir, resumir texto , y clasificar sentimientos, en este caso no interesa hacer un resumen del texto
      entrada = "summarize: " + str(texto)
      #Como tal el modelo no entiende palabras si no números, por lo que debemos codificar el texto
      tokens = tokenizer.encode(entrada, return_tensors="pt", max_length=max_long, truncation=True)
      resumen_ids = model.generate(tokens,
                                   max_length=resumen_max, #Aqui le pasamos el tamaño máximo del resumen en tokens
                                   min_length=resumen_min, #Aqui el mínimo
                                   length_penalty=penalizacion, #Se le agrega una penalización para darle más prioridad a resumenes largos
                                                       #Si fuera de 1, entonces la prioridad serían resumes cortos
                                   num_beams=4, #Este parámetro controla los "caminos" que explora el modelo, con tal de
                                                #generar resumenes con mayor calidad (1=menor calidad, mayor rapidez)
                                   early_stopping=True) #Si el modelo encontro un resumen lo suficientemente bueno entonces
                                                        #se detiene, esto es para no ocupar necesariamente la longitud máxima
      #Como el modelo nos regresa una lista codificada ([0,24,56,264]) tenemos que decodificar dichos numeros para pasarlos
      #a lenguaje natural
      resumen = tokenizer.decode(resumen_ids[0], skip_special_tokens=True)
      resúmenes.append(resumen) #Guardamos el resumen de cada artículo

  return resúmenes
```

Antes de realizar los resumenes, el modelo T5 nos pide darle el largo de la noticia que queremos generar, por defecto se escogió un valor de 100, pero podemos cambiar este valor analizando la longitud del texto original así como de su resumen original.

- Si la noticia tiene una longitud menor a 100, entonces consideramos que la longitud del resumen sea 30 palabras y una penalización igual o menor 1, lo cual favorecería el hacer resumenes cortos.
- Si la noticia tiene una longitud mayor a 100, entonces quedaría la configuración de la longitud del resumen sea de 100 palabras y la penalización de 2.0 para hacer resumenes largos.

No podemos solo aumentar y/o bajar la longitud máxima de los resumenes generados en general porque si la noticia tiene menos palabras de las que queremos generar, el modelo va a empezar a tener errores, o en otro caso, podría empezar a generar resumenes con poca lógica, por eso se considero hacer estos criterios para el cambio de parámetros.

##### Implementación

```python
#Importamos las liberias que contiene los modelos preentrenados
from transformers import BartTokenizer, BartForConditionalGeneration
from tqdm import tqdm

#Igual que con T5 creamos una función para realizar los resumenes con BART
def resumir_textos_bart(textos, resumen_max=100, resumen_min=30, penalizacion=0.0):

  #Aqui también tenemos que escoger que modelo vamos a usar, en este caso usaremos uno que ya esta "fino" y que hace
  #resumenes en inglés
  model_name = 'facebook/bart-large-cnn'
  tokenizer = BartTokenizer.from_pretrained(model_name) #Tokenizamos el texto con ayuda de los módulos de BART
  model = BartForConditionalGeneration.from_pretrained(model_name) #Cargamos el modelo para hacer los resumens
  resúmenes = []
  for texto in tqdm(textos, desc="Generando resúmenes con BART"):

      #Igual que con T5, tenemos que codificar los tokens para que el modelo pueda trabajar con ellos, aparte
      #BART tiene una condicion especial o requisito, y es que este trabajo con tensores, entonces
      #después de hacer la codificación tenemos que devolver un tensor para que el modelo pueda trabajar.
      inputs = tokenizer.encode(texto, return_tensors="pt", max_length=1024, truncation=True) #A diferencia de T5, BART puede trabajar con textos más largos
      # Generar resumen
      resumen_ids = model.generate(
          inputs,
          max_length=resumen_max, #El largo máximo del resumen que se va a generar
          min_length=resumen_min, #El largo mínimo del resumen
          length_penalty=penalizacion, #Penalizamos los resumenes cortos, y damos más importancia a los resumenes largos
          num_beams=4, #Hacemos que el modelo vea más "caminos" (opciones) para generar un resumen mejor
          early_stopping=True #Si ya se encontro un resumen lo sufientemente bueno, entonces el modelo se queda con ese y para.
      )
      # Decodifica la salida igual que con T5
      resumen = tokenizer.decode(resumen_ids[0], skip_special_tokens=True)
      resúmenes.append(resumen)
  return resúmenes
```

```python
#Ahora hacemos los resumenes con BART
resumenes_cortos = resumir_textos_bart(noticias_cortas, resumen_max=30, penalizacion=0.8)
resumenes_largos = resumir_textos_bart(noticias_largas, resumen_max=100, penalizacion=2)

#Convertimos en un dataframe los resumenes generados con los indices originales
resumenes_cortos_df = pd.Series(resumenes_cortos, index=noticias_cortas.index)
resumenes_largos_df = pd.Series(resumenes_largos, index=noticias_largas.index)

#Combinamos los dataset
resumenes_combinados = pd.concat([resumenes_cortos_df, resumenes_largos_df])

#Guardamos los resumenes
df.loc[random_indices, 'resumen_bart'] = resumenes_combinados
```

```python
#Calculamos la métrica de ROUGE entre los resumenes generados y los resumenes originales
#Para ver que tran preciso fue el resumen generado y ver su calidad
def calcular_rouge(lista_generados, lista_referencia):
  """
  Para este caso, ROUGE tiene 3 variantes:
    ROUGE-1	Coincidencia de unigramas (palabras individuales).
    ROUGE-2	Coincidencia de bigramas (pares de palabras consecutivas).
    ROUGE-L	Longest Common Subsequence: la secuencia común más larga.
  Como tenemos algunas noticias cortas y otras largas, optamos por elegir solo la variante 1, donde comparamos palabras
  individuales para encontrar que tan similares son
  """
  scorer = rouge_scorer.RougeScorer(['rouge1'], use_stemmer=True)
  resultados = []
  for generado, referencia in tqdm(zip(lista_generados, lista_referencia), desc="Calculando ROUGE"):
      score = scorer.score(str(referencia), str(generado))
      resultados.append({
          'ROUGE-1_P': score['rouge1'].precision,
          'ROUGE-1_R': score['rouge1'].recall,
          'ROUGE-1_F1': score['rouge1'].fmeasure
      })
      """
      ROUGE te devuelve tres valores que representan lo siguiente:
        - Presición: Mide la proporción de palabras que coinciden del resumen original con respecto al generado.
        - Recall: Mide la proporción de palabras que coinciden del resumen generado con respecto al original.
        - F1: Es la media armónica de precisión y recall.
      Ejemplo: Si consideramos que el resumen generado tiene 15 palabras y el resumen original tiene 25 los valores
      quedarían como:
        - Presición = 10/15 = 0.666
        - Recall = 10/25 = 0.4
        - F1 = 2 * (0.666 * 0.4) / (0.666 + 0.4)
      """
  return pd.DataFrame(resultados)
```

```python
from rouge_score import rouge_scorer
import pandas as pd
from tqdm import tqdm

dfrougeBART = calcular_rouge(dfBART, dforiginal)
dfrougeT5 = calcular_rouge(dfT5, dforiginal)

print("\n\n\nMétrica de ROUGE (T5)")
print(dfrougeT5.mean())
print("\n\n\nMétrica de ROUGE (BART)")
print(dfrougeBART.mean())
```

1. BART supera ligeramente a T5 en todas las métricas:
- En ROUGE-1 y ROUGE-L, BART tiene mejores valores de precision, recall y F1, lo que indica que logra capturar más palabras clave y estructuras del resumen original.

- En ROUGE-2, BART también tiene ventaja, lo que sugiere que genera frases más parecidas a las del resumen original en cuanto a combinaciones de palabras.

2. Diferencias clave:
- Las diferencias no son extremadamente grandes, pero consistentes a favor de BART. Esto implica que ambos modelos son comparables, aunque BART tiende a generar resúmenes más alineados con los originales.

- Las F1-scores (la métrica más representativa en tareas de resumen) muestran un margen de mejora modesto para T5 frente a BART:

  - ROUGE-1 F1: BART (0.476) > T5 (0.465)

  - ROUGE-2 F1: BART (0.297) > T5 (0.278)

  - ROUGE-L F1: BART (0.456) > T5 (0.438)

3. ROUGE-L destaca la ventaja estructural de BART:
- ROUGE-L mide la habilidad del modelo para conservar la estructura y el orden de las ideas. Aquí también BART tiene mejor desempeño, lo cual indica que BART conserva mejor la organización lógica del contenido.

**Conclusión:**

- BART ofrece un mejor rendimiento global en la tarea de resumen automático en comparación con T5, aunque la diferencia no es dramática.

- Si se busca mayor fidelidad al resumen original, especialmente en términos de frases completas y estructura narrativa, BART es una mejor elección.

```python
# Cargar modelo de spaCy
nlp = spacy.load("en_core_web_lg")

def preprocesar_spacy(texto):
    doc = nlp(str(texto))
    return [token.lemma_.lower() for token in doc if token.is_alpha and not token.is_stop]

# Aplicar a la columna Content
df['tokens'] = df['Content'].dropna().apply(preprocesar_spacy)
```

```python
from gensim import corpora, models

# Crear diccionario y corpus
diccionario = corpora.Dictionary(df['tokens'])
corpus = [diccionario.doc2bow(texto) for texto in df['tokens']]

# Entrenar el modelo LDA
lda_modelo = models.LdaModel(
    corpus=corpus,
    id2word=diccionario,
    num_topics=5,
    passes=10,
    random_state=42
)

# Mostrar temas detectados
for idx, topic in lda_modelo.print_topics(num_words=10):
    print(f"Tema {idx}: {topic}")
```

```python
# Función para extraer el tema dominante
def asignar_tema(doc_bow):
    temas = lda_modelo.get_document_topics(doc_bow)
    return max(temas, key=lambda x: x[1])[0] if temas else -1

# Aplicar a cada documento
df['Tema_LDA'] = [asignar_tema(doc) for doc in corpus]
```
